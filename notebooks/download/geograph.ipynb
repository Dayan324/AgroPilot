### Selección Municipios

A continuación se toman como referencia los municipios de Colombia, se seleccionan los correspondientes al departamento de interés y se exportan a .gpkg (Geopackage) para ser utilizados como capa de referencia espacial en otros procesos

```python
import geopandas as gpd
gdf_colombia = gpd.read_file("../../data/shapefiles/Colombia.shp")
# gdf_colombia.head()
```

```python
gdf_colombia.info()
```

```python
gdf_colombia.crs
```

```python
departamento = "Boyacá"
gdf_dpto = gdf_colombia[gdf_colombia["Depto"].eq(departamento)].copy()
gdf_dpto.head()
```

```python
gdf_dpto.to_file("../../data/spatial/municipios.gpkg", driver="GPKG")
```
